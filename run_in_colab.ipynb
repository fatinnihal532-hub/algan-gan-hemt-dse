## Try your own device

Build any barrier stack / geometry and read off its full set of extracted
figures of merit.

```python
from src.materials import AlGaN
from src.charge_control import BarrierStack
from src.current_model import HEMTDevice, HEMTGeometry
from src.extraction import extract_metrics

stack = BarrierStack(AlGaN(x=0.28), thickness_m=18e-9)
geom = HEMTGeometry(gate_length_m=0.2e-6, gate_drain_spacing_m=3e-6)
dev = HEMTDevice(stack, geom)

m = extract_metrics(dev)
for field, value in m.__dict__.items():
    print(f"{field:28s} {value:.4g}")
```
